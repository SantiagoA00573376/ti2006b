## Research Question
What is the average transaction amount, and how many transactions are potentially fraudulent?

## Data Provenance
This exercise uses synthetic financial transaction data created in Python. No real or sensitive information is used.

## Expected Output
Calculate the average transaction amount and count the number of potentially fraudulent transactions.

```python
import sys
import pandas as pd

print("Python version:", sys.version)
print("Pandas version:", pd.__version__)
```

```python
# Create synthetic financial transaction data

data = {
    "Transaction_ID": [1, 2, 3, 4, 5],
    "Amount": [150, 2500, 80, 7500, 320],
    "Fraud": [0, 0, 0, 1, 0]
}

df = pd.DataFrame(data)

print("Synthetic Financial Transactions")
display(df)
```

```python

# Analyze synthetic financial transactions

average_amount = df["Amount"].mean()
fraud_count = df["Fraud"].sum()

print("Average Transaction Amount:", average_amount)
print("Fraudulent Transactions:", fraud_count)
```

Conclusión

El análisis se realizó utilizando cinco transacciones financieras con datos sintéticos, es decir, información ficticia creada únicamente para esta actividad.

El monto promedio de las transacciones fue de 2,110 unidades monetarias. Además, se identificó una transacción marcada como fraudulenta, lo que representa el 20% del total.

Esta actividad demuestra cómo Python y la librería Pandas permiten organizar y analizar información financiera de manera sencilla.

Sin embargo, debido a que los datos son ficticios y la muestra es pequeña, los resultados no representan patrones reales de fraude financiero.
